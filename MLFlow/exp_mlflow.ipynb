```python
import mlflow
import os

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
```

```python
mlflow.set_experiment("Loan_Status_Experiment")
```

```python
train_df = pd.read_csv("data.csv")
train_df.head()
```

```python
train_df.Loan_Status.value_counts()
```

```python
#Let's binary encode Gender, Married, Loan Status

train_df['Gender'] = train_df['Gender'].map({'Male': 1, 'Female': 0})
train_df['Married'] = train_df['Married'].map({'Yes': 1, 'No': 0})
train_df['Loan_Status'] = train_df['Loan_Status'].map({'Y': 1, 'N': 0})
```

```python
train_df.info()
```

```python
train_df.isnull( ).sum()
```

```python
#dropping all the missing values 
train_df = train_df.dropna()
train_df.isnull().sum()

#Don't drop features off data off in production
```

#Let's create simple features

```python
feature_columns = ['Gender','Married', 'ApplicantIncome', 'LoanAmount', 'Credit_History', 'CoapplicantIncome', 'Loan_Amount_Term', 'Credit_History']
X = train_df[feature_columns]
y = train_df.Loan_Status
```

```python
mlflow.log_param("feature_columns", feature_columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, random_state=5)
```

```python
#Let's begin model training
```

```python
from sklearn.ensemble import RandomForestClassifier


#initially test for 
# depth = 6
# criteria = 'gini' #gini is accurate and faster than entropy and is default
# n_esti = 1000


depth = 10
criteria = 'entropy'
n_esti = 10

#will the bias increase of decrease if the number of estimators increase? interview question  

mlflow.log_param("max_depth", depth)
mlflow.log_param("criterion", criteria)
mlflow.log_param("n_estimators", n_esti)
mlflow.log_param("model_name", "RandomForestClassifier")

#mlflow.log_param("model_name", "XgboostClassifier")
#mlflow.log_param({
#"max_depth": depth,
#"criterion": criteria,
#"n_estimators": n_esti
#})


model = RandomForestClassifier(max_depth=depth, criterion=criteria, n_estimators=n_esti)
model.fit(X_train, y_train)

#import xgboost as xgb
#model = xgb.XBGClassifier(max_depth=depth, n_estimators=n_esti)
model.fit(X_train, y_train)


#which can be paralyzed and which cannot be paralyzed? interview question
```

```python
from sklearn.metrics import accuracy_score

pred_val = model.predict(X_val)
val_acc_score = accuracy_score(y_val, pred_val)
print(f"Training Accuracy: {model.score(X_train, y_train)}, Validation Accuracy: {val_acc_score}")
```

```python
mlflow.log_metric("train_accuracy", model.score(X_train, y_train))
mlflow.log_metric("val_accuracy", val_acc_score)
```

```python
mlflow.end_run()
```
